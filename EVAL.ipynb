```python
from deepeval.test_case import LLMTestCase, LLMTestCaseParams
from deepeval.metrics import GEval

# Example inputs from your pipeline
literary_question = "What central themes does Jane Austen explore in 'Pride and Prejudice'?"
actual_output = """
The text explores several central themes, including love and marriage, social class, pride and prejudice, reputation, and individual growth.
These themes reflect the social context of Austen's time and provide insight into human behavior and relationships.
"""

# Build the test case
test_case = LLMTestCase(
    input=literary_question,
    actual_output=actual_output
)

# Define a customized GEval metric for coherence and thematic relevance
coherence_metric = GEval(
    name="Literary Coherence",
    criteria=(
        "The response must be coherent, logically structured, and maintain a consistent "
        "literary tone. The analysis should accurately reflect the context and themes "
        "of the author’s work."
    ),
    evaluation_params=[LLMTestCaseParams.ACTUAL_OUTPUT],
)

# Run the measurement
coherence_metric.measure(test_case)

print(f"Score: {coherence_metric.score}")
print(f"Reason: {coherence_metric.reason}")
```
